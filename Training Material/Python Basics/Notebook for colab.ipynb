# 🔤 Understanding Python Data Types

In Python, you don’t need to tell the computer what type of data a variable will hold.  
Python figures it out **dynamically** (automatically) when you assign a value.

For example:
- Text like `"gpt-4"` is stored as a **string** (`str`)
- Numbers with decimals like `0.7` are stored as a **floating-point number** (`float`)
- Whole numbers like `200` are stored as an **integer** (`int`)
- Values like `True` or `False` are stored as **booleans** (`bool`)

👉 We can check the type of any variable using the built-in `type()` function.

In the next cell, we’ll:
- Create variables with different kinds of values
- Use `print(type(...))` to see what Python thinks each variable’s type is

```python
model_name = "gpt-4"
temperature = 0.7
max_tokens = 200
is_active = True

print(type(model_name))
print(type(temperature))
print(type(max_tokens))
print(type(is_active))
```

# 📦 Python Data Structures: List, Dictionary, Tuple, and Set

Python gives us different ways to organize and store data.  
Each structure has its own rules and use cases:

- **List**
  - Ordered collection (items keep their position)
  - Mutable (you can change it)
  - Allows duplicates
  - Example: a list of chat messages

- **Dictionary**
  - Stores data as **key-value pairs**
  - Keys must be unique
  - Great for configurations or labeled data
  - Example: model settings like `temperature` or `max_tokens`

- **Tuple**
  - Ordered collection (like a list)
  - Immutable (cannot be changed once created)
  - Allows duplicates
  - Example: coordinates (latitude, longitude)

- **Set**
  - Unordered collection
  - Mutable (you can add/remove items)
  - Does **not** allow duplicates
  - Example: unique roles in a conversation

👉 In the next cell, we’ll create examples of each and print them out to see how they look in Python.

```python
lista=[{}, {}, {}]
mydict={"key": "value", "age": 25, "city": "Karachi"}
mytuple=(1, 2, 3, 4, 5)
myset={1, 2, 3, 4, 5}
```

# 🔄 Introduction to Loops in Python

Sometimes we want to repeat actions multiple times without writing the same code again and again.  
That’s where **loops** come in!

Python has two main types of loops:

- **For Loop**
  - Used when you want to go through (iterate over) a collection of items.
  - Example: going through a list of prompts one by one.
  - The loop stops automatically when it reaches the end of the list.

- **While Loop**
  - Used when you want to keep repeating something **as long as a condition is true**.
  - Example: keep printing a message until a counter reaches 3.
  - You must update the condition (like increasing the counter) to avoid an infinite loop.

👉 Think of loops as a way to tell the computer:  
*"Do this task again and again until I say stop."*

In the next cell, we’ll see:
- A **for loop** that prints each prompt in a list.  
- A **while loop** that runs until a counter reaches 3.

```python
# prompts = [
#     "What is an LLM?",
#     "What are embeddings?",
#     "What is tokenization?"
# ]

# for x in prompts:
#     print("Sending prompt:", x)


counter = 0

# While loop will keep running until counter reaches 3
while counter < 5:
    print("This is loop iteration number:", counter)
    counter += 1
    
print('outside while loop')
```

# ⚖️ Conditional Statements in Python

Sometimes we want our program to make decisions.  
Conditional statements let us run different pieces of code depending on certain conditions.

Python uses:

- **`if`** → checks a condition, runs code if it’s true  
- **`elif`** (else if) → checks another condition if the first one was false  
- **`else`** → runs code if none of the above conditions are true  

👉 Think of it like asking questions:
- *If it’s raining → take an umbrella*  
- *Else if it’s sunny → wear sunglasses*  
- *Else → just go outside normally*

This is how computers make choices!

```python
temperature = int(input("Enter a temperature value"))

if temperature > 35:
    print("It's really hot outside!")
elif temperature > 25:
    print("The weather is warm.")
else:
    print("It's nice and cool today.")
```

# 🛠️ Functions, Reusability, and Type Checking

Functions let us **group code into reusable blocks**.  
Instead of writing the same logic again and again, we can put it inside a function and call it whenever we need.

Benefits of functions:
- **Reusability** → write once, use many times
- **Clarity** → code is easier to read and understand
- **Validation** → we can specify what type of input/output we expect

👉 In Python, we can use **type hints** to show what kind of data a function should take and return:
- `user_input: str` means the function expects a string as input
- `-> str` means the function will return a string

This helps us (and others) understand the function better, and tools like editors can warn us if we use the wrong type.

```python
def format_prompt(user_input: str) -> str:
    """
    This function takes user input (a topic)
    and returns a formatted string prompt.
    """
    return f"Explain simply: {user_input}"



result = format_prompt(input("Enter a topic you want to learn about: "))
print(result)
```

# 🔢 Introduction to NumPy

NumPy (Numerical Python) is one of the most important libraries for scientific computing in Python.  
It makes working with numbers, arrays, and mathematical operations **fast and efficient**.

Why NumPy?
- Python lists are flexible but slow for math-heavy tasks.
- NumPy arrays are optimized for numerical operations and can handle large datasets quickly.
- It provides built-in functions for **linear algebra, statistics, random numbers, reshaping arrays, and more**.

👉 Think of NumPy as a **super-powered calculator** for Python.

In the next cell, we’ll:
- Create two vectors (arrays)
- Perform a **dot product** (a common linear algebra operation)
- Print the result

We’ll also highlight some common NumPy functionalities like array creation, math operations, statistics, and random numbers.

```python
import numpy as np

# Create two vectors (arrays)
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

# Dot product of two vectors
dot_product = np.dot(vector_a, vector_b)

print("Dot product:", dot_product)

# Common NumPy functionalities (examples)
print("Zeros array:", np.zeros(3))
print("Ones array:", np.ones(3))
print("Range array:", np.arange(1, 10, 2))
print("Mean of vector_a:", np.mean(vector_a))
print("Random number:", np.random.rand())
```

# 🧾 Introduction to Pandas

Pandas is one of the most popular Python libraries for **data analysis and manipulation**.

In the next cell, we’ll:
- Create a small dataset (dictionary of prompts and response lengths)
- Convert it into a **DataFrame**
- Print it out to see how Pandas organizes data

# 🔑 Introduction to dotenv and Environment Variables

When working with APIs (like OpenAI), we often need to use **secret keys**.  
We don’t want to hardcode these keys directly into our code for security reasons.  
Instead, we store them in a hidden file called `.env` and load them when needed.

👉 The `python-dotenv` library helps us:
- Store secrets in a `.env` file
- Load them into our program safely
- Access them using `os.getenv("KEY_NAME")`

This way, our code stays clean and secure.

```python
import os
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Access the API key
api_key = os.getenv("OPENAI_API_KEY")

print("API Key loaded:", api_key is not None)
```
