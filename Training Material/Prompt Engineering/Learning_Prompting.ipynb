```python
# Import libraries
import os
from dotenv import load_dotenv
from openai import AzureOpenAI
```

```python
# Load environment variables
load_dotenv()
```

```python
# Configure Azure OpenAI client
client = AzureOpenAI(
    api_key=os.getenv("AZURE_OPENAI_KEY"),
    api_version="2024-05-01-preview",
    azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT")
)
```

## Zero-Shot Learning

**Definition:**  
Zero-shot learning means we ask the model to perform a task without giving any examples.  
We rely on the model’s pre-trained knowledge.

**Why we need it:**  
Useful when we don’t have examples, or when the task is simple and general.


```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Translate 'Hello, how are you?' into French."}
    ]
)
print(response.choices[0].message.content)
```

## One-Shot Learning

**Definition:**  
One-shot learning means we provide **one example** of input-output before asking the model to solve a new case.

**Why we need it:**  
Helps the model understand the format or style we want.


```python
prompt = """
Example:
English: Good morning
French: Bonjour

Now translate:
English: How are you?
French:
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}]
)
print(response.choices[0].message.content)
```

## Few-Shot Learning

**Definition:**  
Few-shot learning means we provide **multiple examples** of input-output pairs.  
The model learns the pattern and applies it to new inputs.

**Why we need it:**  
Improves accuracy when the task is complex or requires a specific style.


```python
prompt = """
Examples:
English: Good morning
French: Bonjour

English: Good night
French: Bonne nuit

English: Thank you
French: Merci

Now translate:
English: See you tomorrow
French:
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}]
)
print(response.choices[0].message.content)
```

## Chain-of-Thought Prompting

**Definition:**  
Chain-of-thought prompting encourages the model to **show its reasoning step by step** before giving the final answer.

**Why we need it:**  
Useful for math, logic, or reasoning tasks where transparency is important.


```python
prompt = """
Question: If I have 3 apples and buy 2 more, how many apples do I have?
Answer: Let's think step by step.
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}]
)
print(response.choices[0].message.content)
```

## Tree-of-Thought Prompting

**Definition:**  
Tree-of-thought prompting explores **multiple reasoning paths** before choosing the best answer.  
It’s like brainstorming different solutions and then selecting the most logical one.

**Why we need it:**  
Helps with complex problem-solving where multiple approaches are possible.


```python
prompt = """
Question: What is the best way to reduce traffic in a busy city?
Answer: Consider multiple possible solutions, evaluate them, and then choose the best one.
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt}]
)
print(response.choices[0].message.content)
```
